## 4. Apparent-Magnitude Bin Split

Production VDP maps are generated separately by apparent-magnitude bin. This section attaches `mag_app`, counts the bins, and can restrict the cloning test to one bin (optional).

```python
scorer = population_scorers[POP_TO_TEST]

mag_app = vdp.compute_apparent_magnitude_for_population(
    df=source_df_sample,
    obstime_str=OBSTIME_STR,
    scorer=scorer,
    chunk=50_000,
    show_progress=True,
)

source_df_with_mag = source_df_sample.copy()
source_df_with_mag['mag_app'] = mag_app

def assign_mag_bin(mag):
    if not np.isfinite(mag):
        return None
    for mb in vdp.DEFAULT_MAG_BINS:
        if mb['mag_min'] <= mag < mb['mag_max']:
            return mb['label']
    return None

source_df_with_mag['mag_bin_label'] = [assign_mag_bin(m) for m in source_df_with_mag['mag_app']]

bin_summary = []
for mb in vdp.DEFAULT_MAG_BINS:
    in_bin = source_df_with_mag['mag_bin_label'] == mb['label']
    bin_summary.append({
        'label': mb['label'],
        'mag_min': mb['mag_min'],
        'mag_max': mb['mag_max'],
        'n_sample': int(in_bin.sum()),
    })
bin_summary.append({
    'label': 'outside_bins',
    'mag_min': np.nan,
    'mag_max': np.nan,
    'n_sample': int(source_df_with_mag['mag_bin_label'].isna().sum()),
})

pd.DataFrame(bin_summary)
```

## 9. Build a Test Density Map

```python
x_grid, y_grid, X0, Y0, grid_points = vdp.make_default_grid(
    grid_lim=GRID_LIM,
    grid_step=GRID_STEP,
)

pts = clone_visible_df[['vlam', 'vbeta']].to_numpy(dtype=float)
good_pts = np.isfinite(pts).all(axis=1)
pts = pts[good_pts]

print('Finite velocity points:', len(pts))

if len(pts) < 3:
    raise ValueError('Too few visible points for density-map diagnostic. Increase MAX_SOURCE_OBJECTS or CLONE_FACTOR_TEST.')

tree = cKDTree(pts)
k_eff = min(K_MAP_TEST, len(pts) - 1)
density_flat = vdp.evaluate_density_map_full_posterior_2d(
    tree,
    grid_points,
    k=k_eff,
    n_d0_grid=N_D0_GRID_TEST,
    show_progress=True,
)
density_raw = density_flat.reshape(X0.shape) / CLONE_FACTOR_TEST

support_count = vdp.make_support_count_map(
    pts[:, 0], pts[:, 1], x_grid, y_grid, clone_factor=CLONE_FACTOR_TEST,
)

print('k_eff:', k_eff)
print('raw density max:', float(np.nanmax(density_raw)))
print('support max:', float(np.nanmax(support_count)))
```

## 10. Apply the Production-Style Smoothing Passes

```python
density_smoothed = density_raw.copy()
support_for_smoothing = support_count * CLONE_FACTOR_TEST if SCALE_SUPPORT_BY_CLONE_FACTOR else support_count

for pass_cfg in SMOOTH_PASSES:
    density_smoothed = vdp.smooth_density_map_by_support(
        density_smoothed,
        support_for_smoothing,
        support_threshold=pass_cfg['support_threshold'],
        sigma_pixels=pass_cfg['sigma_pixels'],
        truncate_sigma=pass_cfg['truncate_sigma'],
    )

density_delta = density_smoothed - density_raw

print('smoothing passes:', SMOOTH_PASSES)
print('support scaled by clone factor:', SCALE_SUPPORT_BY_CLONE_FACTOR)
print('smoothed density max:', float(np.nanmax(density_smoothed)))
print('median |delta|:', float(np.nanmedian(np.abs(density_delta))))
```

## 11. Density Smoothing Difference Plot

```python
def density_log_map(density, floor=1e-30):
    data = np.log10(np.maximum(density, floor))
    finite = np.isfinite(data)
    if not np.any(finite):
        return data, -1.0, 1.0
    vmax = float(np.nanmax(data[finite]))
    return data, vmax - 5.0, vmax


raw_log, vmin, vmax = density_log_map(density_raw)
smooth_log, _, _ = density_log_map(density_smoothed)

lim = np.nanpercentile(np.abs(density_delta), 99.5)
if not np.isfinite(lim) or lim == 0:
    lim = 1.0

fig, axes = plt.subplots(1, 3, figsize=(15, 4.5), constrained_layout=True)

im0 = axes[0].pcolormesh(X0, Y0, raw_log, shading='auto', cmap='viridis', vmin=vmin, vmax=vmax)
axes[0].set_title('loaded/raw density')
fig.colorbar(im0, ax=axes[0], label='log10 density')

im1 = axes[1].pcolormesh(X0, Y0, smooth_log, shading='auto', cmap='viridis', vmin=vmin, vmax=vmax)
axes[1].set_title('smoothed density')
fig.colorbar(im1, ax=axes[1], label='log10 density')

im2 = axes[2].pcolormesh(X0, Y0, density_delta, shading='auto', cmap='coolwarm', vmin=-lim, vmax=lim)
axes[2].set_title('smoothed - raw')
fig.colorbar(im2, ax=axes[2], label='density delta')

for ax in axes:
    ax.set_xlim(*GRID_LIM)
    ax.set_ylim(*GRID_LIM)
    ax.set_xlabel(r'$v_\lambda$ (deg/day)')
    ax.set_ylabel(r'$v_\beta$ (deg/day)')

fig.suptitle(f'{POP_TO_TEST}: density smoothing diagnostic')
plt.show()
```

## 13. NEO-only Gaussian-mixture cloner v2

This section completes the advisor prototype into a VDP-ready NEO cloner.
It is intentionally **NEO-only**: MBA, TNO, and Trojan cloning should stay
on the existing conditional K|M path.

The GMM is fit in a physically safer feature space:

- `log(a)` instead of raw `a`
- `q = a(1-e)` as the NEO-defining perihelion coordinate
- sine/cosine pairs for angular variables
- mean anomaly at the map epoch, then reconstruct `t_p`
- `H` sampled separately from the empirical/source NEO distribution

The output is a normal orbital-element DataFrame with
`a, e, i, node, argperi, t_p, H`, where `H` is attached after GMM sampling so the observed faint-end distribution is preserved. It can go straight through
`vdp.build_visible_subset_dataframe(...)` and the same density-map
diagnostics above.

```python
from sklearn.mixture import GaussianMixture
from sklearn.preprocessing import StandardScaler
from astropy.time import Time

GMM_NCOMP_TEST = 80
GMM_TARGET_CLONES = min(250_000, max(10_000, len(source_df) * CLONE_FACTOR_TEST))
GMM_MAX_ATTEMPTS = 12
GMM_OVERSAMPLE = 1.8

print('GMM target clones:', GMM_TARGET_CLONES)
print('GMM components:', GMM_NCOMP_TEST)
print('Training rows:', len(source_df))
print('H sampling: empirical source distribution, not a GMM feature')
```

```python
def _as_q(df):
    if 'q' in df.columns:
        return df['q'].to_numpy(dtype=float)
    return df['a'].to_numpy(dtype=float) * (1.0 - df['e'].to_numpy(dtype=float))


def _mean_anomaly_rad(df, obstime_str):
    t_obs = Time(obstime_str, scale='tdb')
    a = df['a'].to_numpy(dtype=float)
    tp = df['t_p'].to_numpy(dtype=float)
    n_rad_day = np.sqrt(vdp.MU_SUN / ((a * vdp.AU_KM) ** 3)) * 86400.0
    return np.mod((t_obs.mjd - tp) * n_rad_day, 2.0 * np.pi)


def _angle_deg_from_sincos(s, c):
    return np.degrees(np.mod(np.arctan2(s, c), 2.0 * np.pi))


def _inclination_deg_from_sincos(s, c):
    angle = _angle_deg_from_sincos(s, c)
    return np.where(angle <= 180.0, angle, 360.0 - angle)


def _sample_empirical_h(train, n, rng):
    h_source = train['H'].to_numpy(dtype=float)
    h_source = h_source[np.isfinite(h_source)]
    if len(h_source) == 0:
        raise ValueError('No finite H values available for empirical H sampling.')
    return rng.choice(h_source, size=n, replace=True)


def _neo_gmm_feature_matrix(df, obstime_str):
    a = df['a'].to_numpy(dtype=float)
    q = _as_q(df)
    inc = np.deg2rad(df['i'].to_numpy(dtype=float))
    node = np.deg2rad(df['node'].to_numpy(dtype=float))
    argp = np.deg2rad(df['argperi'].to_numpy(dtype=float))
    M = _mean_anomaly_rad(df, obstime_str)

    X = np.column_stack([
        np.log(a),
        q,
        np.sin(inc), np.cos(inc),
        np.sin(node), np.cos(node),
        np.sin(argp), np.cos(argp),
        np.sin(M), np.cos(M),
    ])
    names = [
        'log_a', 'q',
        'sin_i', 'cos_i',
        'sin_node', 'cos_node',
        'sin_argperi', 'cos_argperi',
        'sin_M', 'cos_M',
    ]
    return X, names


def _unpack_neo_gmm_features(X, obstime_str):
    t_obs = Time(obstime_str, scale='tdb')
    log_a = X[:, 0]
    q = X[:, 1]
    a = np.exp(log_a)
    e = 1.0 - q / a

    inc = _inclination_deg_from_sincos(X[:, 2], X[:, 3])
    node = _angle_deg_from_sincos(X[:, 4], X[:, 5])
    argp = _angle_deg_from_sincos(X[:, 6], X[:, 7])
    M = np.deg2rad(_angle_deg_from_sincos(X[:, 8], X[:, 9]))

    n_rad_day = np.sqrt(vdp.MU_SUN / ((a * vdp.AU_KM) ** 3)) * 86400.0
    t_p = t_obs.mjd - M / n_rad_day

    return pd.DataFrame({
        'a': a,
        'e': e,
        'i': inc,
        'node': node,
        'argperi': argp,
        't_p': t_p,
        'q': q,
        'M_obs_deg': np.degrees(M) % 360.0,
    })


def clone_neo_gmm_orbits(
    df,
    n_clones,
    obstime_str,
    n_components=80,
    random_state=42,
    oversample=1.8,
    max_attempts=12,
    verbose=True,
):
    rng = np.random.default_rng(random_state)
    train = df.copy()
    q_train = _as_q(train)
    finite_train = (
        np.isfinite(train['a'].to_numpy(float)) &
        np.isfinite(train['e'].to_numpy(float)) &
        np.isfinite(train['i'].to_numpy(float)) &
        np.isfinite(train['node'].to_numpy(float)) &
        np.isfinite(train['argperi'].to_numpy(float)) &
        np.isfinite(train['t_p'].to_numpy(float)) &
        np.isfinite(train['H'].to_numpy(float)) &
        np.isfinite(q_train) &
        (train['a'].to_numpy(float) > 0) &
        (q_train >= 0) &
        (q_train < 1.3) &
        (train['e'].to_numpy(float) >= 0) &
        (train['e'].to_numpy(float) < 1)
    )
    train = train.loc[finite_train].copy()
    if len(train) < 10:
        raise ValueError(f'Too few valid NEO training rows for GMM: {len(train)}')

    X, feature_names = _neo_gmm_feature_matrix(train, obstime_str)
    finite = np.isfinite(X).all(axis=1)
    X = X[finite]
    train = train.iloc[np.where(finite)[0]].copy()

    scaler = StandardScaler()
    X_scaled = scaler.fit_transform(X)

    n_components_eff = min(int(n_components), max(1, len(X_scaled) // 5), len(X_scaled))
    gmm = GaussianMixture(
        n_components=n_components_eff,
        covariance_type='full',
        reg_covar=1e-6,
        random_state=random_state,
        max_iter=300,
        n_init=2,
    )
    gmm.fit(X_scaled)

    i_min, i_max = 0.0, min(180.0, np.nanmax(train['i'].to_numpy(float)) + 5.0)
    a_min, a_max = np.nanpercentile(train['a'].to_numpy(float), [0.1, 99.9])
    a_min = max(0.05, 0.5 * a_min)
    a_max = 1.5 * a_max

    accepted = []
    total_drawn = 0

    for attempt in range(max_attempts):
        remaining = n_clones - sum(len(x) for x in accepted)
        if remaining <= 0:
            break
        batch = max(1000, int(np.ceil(remaining * oversample)))
        Xs, labels = gmm.sample(batch)
        Xraw = scaler.inverse_transform(Xs)
        sim = _unpack_neo_gmm_features(Xraw, obstime_str)
        sim['H'] = _sample_empirical_h(train, len(sim), rng)
        total_drawn += len(sim)

        valid = (
            np.isfinite(sim[['a', 'e', 'i', 'node', 'argperi', 't_p', 'H', 'q']]).all(axis=1) &
            (sim['a'] > a_min) & (sim['a'] < a_max) &
            (sim['q'] >= 0.0) & (sim['q'] < 1.3) &
            (sim['e'] >= 0.0) & (sim['e'] < 1.0) &
            (sim['i'] >= i_min) & (sim['i'] <= i_max)
        )
        sim = sim.loc[valid, ['a', 'e', 'i', 'node', 'argperi', 't_p', 'H', 'q', 'M_obs_deg']]
        if len(sim):
            accepted.append(sim)
        if verbose:
            got = sum(len(x) for x in accepted)
            print(f'GMM attempt {attempt + 1}: accepted {len(sim):,}/{batch:,}; total {got:,}/{n_clones:,}')

    if not accepted:
        raise RuntimeError('GMM produced no valid NEO clones.')

    out = pd.concat(accepted, ignore_index=True).head(n_clones)
    diagnostics = {
        'feature_names': feature_names,
        'H_sampling': 'empirical_source_distribution',
        'n_training': len(train),
        'n_components_eff': n_components_eff,
        'total_drawn': total_drawn,
        'acceptance_fraction': len(out) / max(total_drawn, 1),
        'converged': bool(gmm.converged_),
        'lower_bound': float(gmm.lower_bound_),
    }
    return out, gmm, scaler, diagnostics
```

```python
gmm_training_df = source_df.copy()
if POP_TO_TEST != 'NEO':
    raise ValueError('This GMM test section is intentionally NEO-only. Set POP_TO_TEST = "NEO".')

gmm_raw_clone_df, gmm_model, gmm_scaler, gmm_diag = clone_neo_gmm_orbits(
    gmm_training_df,
    n_clones=GMM_TARGET_CLONES,
    obstime_str=OBSTIME_STR,
    n_components=GMM_NCOMP_TEST,
    random_state=RNG_SEED,
    oversample=GMM_OVERSAMPLE,
    max_attempts=GMM_MAX_ATTEMPTS,
)

print('GMM diagnostics:')
for k, v in gmm_diag.items():
    print(f'  {k}: {v}')

print('\nGMM raw clone table:', gmm_raw_clone_df.shape)
gmm_raw_clone_df.head()
```

## 16. Build the same density/smoothing diagnostic for GMM clones

```python
def build_test_density_from_visible(visible_df, clone_factor_for_downweight, label):
    pts = visible_df[['vlam', 'vbeta']].to_numpy(dtype=float)
    good_pts = np.isfinite(pts).all(axis=1)
    pts = pts[good_pts]
    if len(pts) < 3:
        raise ValueError(f'Too few finite velocity points for {label}: {len(pts)}')

    tree = cKDTree(pts)
    k_eff = min(K_MAP_TEST, len(pts) - 1)
    density_flat = vdp.evaluate_density_map_full_posterior_2d(
        tree,
        grid_points,
        k=k_eff,
        n_d0_grid=N_D0_GRID_TEST,
        show_progress=True,
    )
    density_raw_local = density_flat.reshape(X0.shape) / clone_factor_for_downweight
    support_count_local = vdp.make_support_count_map(
        pts[:, 0], pts[:, 1], x_grid, y_grid, clone_factor=clone_factor_for_downweight,
    )

    density_smoothed_local = density_raw_local.copy()
    support_for_smoothing_local = support_count_local * clone_factor_for_downweight if SCALE_SUPPORT_BY_CLONE_FACTOR else support_count_local
    for pass_cfg in SMOOTH_PASSES:
        density_smoothed_local = vdp.smooth_density_map_by_support(
            density_smoothed_local,
            support_for_smoothing_local,
            support_threshold=pass_cfg['support_threshold'],
            sigma_pixels=pass_cfg['sigma_pixels'],
            truncate_sigma=pass_cfg['truncate_sigma'],
        )

    return {
        'label': label,
        'pts': pts,
        'k_eff': k_eff,
        'density_raw': density_raw_local,
        'density_smoothed': density_smoothed_local,
        'support_count': support_count_local,
    }

# The GMM is not a per-parent clone factor in the strict sense; use the
# effective generated/source ratio so density normalization remains
# comparable in plots.
GMM_EFFECTIVE_CLONE_FACTOR = max(1, int(round(len(gmm_raw_clone_df) / max(len(gmm_training_df), 1))))
print('GMM effective clone factor:', GMM_EFFECTIVE_CLONE_FACTOR)

gmm_density = build_test_density_from_visible(
    gmm_clone_visible_df,
    clone_factor_for_downweight=GMM_EFFECTIVE_CLONE_FACTOR,
    label='NEO GMM',
)
print('GMM k_eff:', gmm_density['k_eff'])
print('GMM raw density max:', float(np.nanmax(gmm_density['density_raw'])))
print('GMM smoothed density max:', float(np.nanmax(gmm_density['density_smoothed'])))
```

```python
def _usable_density_map(value, expected_shape):
    try:
        arr = np.asarray(value, dtype=float)
    except Exception:
        return False
    return arr.ndim == 2 and arr.shape == expected_shape and np.isfinite(arr).any()


def _as_density_map(value, name, expected_shape):
    arr = np.asarray(value, dtype=float)
    if arr.ndim != 2 or arr.shape != expected_shape:
        raise ValueError(f'{name} has shape {arr.shape}, expected {expected_shape}')
    return arr

# In some run orders, density_raw/density_smoothed can be stale notebook-state
# objects instead of the 2D arrays from section 9/10. Rebuild K|M maps from the
# current visible clone table if that happened.
expected_shape = X0.shape
if (
    'density_raw' not in globals() or
    'density_smoothed' not in globals() or
    not _usable_density_map(density_raw, expected_shape) or
    not _usable_density_map(density_smoothed, expected_shape)
):
    print('Rebuilding K|M density maps for this comparison cell...')
    km_density = build_test_density_from_visible(
        clone_visible_df,
        clone_factor_for_downweight=CLONE_FACTOR_TEST,
        label='K|M',
    )
    km_raw = km_density['density_raw']
    km_smoothed = km_density['density_smoothed']
else:
    km_raw = _as_density_map(density_raw, 'density_raw', expected_shape)
    km_smoothed = _as_density_map(density_smoothed, 'density_smoothed', expected_shape)

gmm_raw = _as_density_map(gmm_density['density_raw'], "gmm_density['density_raw']", expected_shape)
gmm_smoothed = _as_density_map(gmm_density['density_smoothed'], "gmm_density['density_smoothed']", expected_shape)

fig, axes = plt.subplots(2, 3, figsize=(16, 9), constrained_layout=True)

km_maps = [km_raw, km_smoothed, km_smoothed - km_raw]
gmm_maps = [gmm_raw, gmm_smoothed, gmm_smoothed - gmm_raw]
titles = ['raw density', 'smoothed density', 'smoothed - raw']

for row, (maps, row_label) in enumerate([(km_maps, 'K|M'), (gmm_maps, 'GMM')]):
    for col, (arr, title) in enumerate(zip(maps, titles)):
        ax = axes[row, col]
        if title in ('raw density', 'smoothed density'):
            log_arr, log_vmin, log_vmax = density_log_map(arr)
            im = ax.imshow(
                log_arr,
                origin='lower',
                extent=[x_grid[0], x_grid[-1], y_grid[0], y_grid[-1]],
                aspect='auto',
                cmap='viridis',
                vmin=log_vmin,
                vmax=log_vmax,
            )
        else:
            im = ax.imshow(
                arr,
                origin='lower',
                extent=[x_grid[0], x_grid[-1], y_grid[0], y_grid[-1]],
                aspect='auto',
                cmap='coolwarm',
            )
        ax.set_title(f'{row_label}: {title}')
        ax.set_xlabel(r'$v_\lambda$')
        ax.set_ylabel(r'$v_\beta$')
        fig.colorbar(im, ax=ax)

fig.suptitle('Density-map diagnostic: current K|M vs NEO GMM')
plt.show()
```

Use this notebook to answer these questions before touching production
map generation:

1. Does the GMM reproduce the 1D and 2D NEO orbital-element distributions
   without obvious unphysical artifacts?
2. Does the GMM produce visible velocity support in regions where K|M
   leaves holes?
3. Does it avoid creating broad artificial NEO density that would raise
   MBA contamination?
4. Does the GMM result look stable when changing `GMM_NCOMP_TEST`,
   `MAX_SOURCE_OBJECTS`, and `GMM_TARGET_CLONES`?

add a `cloner='neo_gmm'` option for the NEO population and leave
