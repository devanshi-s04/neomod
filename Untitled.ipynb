```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df  = pd.read_hdf("s3m_orbits.h5")
print(df.head())
```

```python

plt.figure(figsize=(7, 6))
sc = plt.scatter(df["e"], df["q"], c=df["inc"], cmap="plasma", s=12, alpha=0.7)
plt.xlabel("Eccentricity (e)")
plt.ylabel("Perihelion Distance (q) [AU]")
plt.title("q vs e Colored by Inclination")
plt.colorbar(sc, label="Inclination (deg)")
plt.grid(True)
plt.show()

```

```python
df["a"] = df["q"] / (1 - df["e"])
```

```python

plt.figure(figsize=(7, 6))
sc = plt.scatter(df["a"], df["e"], c=df["inc"], cmap="plasma", s=12, alpha=0.7)
plt.xlabel("Semimajor Axis (a) [AU]")
plt.ylabel("Eccentricity (e)")
plt.title("a vs e Colored by Inclination")
plt.colorbar(sc, label="Inclination (deg)")
plt.grid(True)
plt.show()
```
